## Symbolic derivatives

`Symbolics` supplies the symbolic variable and the differentiation operator. Two forms are
useful: `Differential(x)` builds an unapplied operator (`expand_derivatives` evaluates it),
while `Symbolics.derivative` does both in one step.

## The difference quotient, from first principles

The page derives $f'(x) = 2x$ for $f(x) = x^2$ by expanding $(x+h)^2 - x^2$, cancelling the
$h$, and letting $h \to 0$. Reproducing that here is a good check on the algebra — and on
the tooling, since a raw $0/0$ quotient must be **expanded and cancelled before** any limit
is taken.

## The tangent line

$t(\bar{x}) = f'(x)\,(\bar{x} - x) + f(x)$. `CalculusWithJuliaSquared` exports `tangent`
and `secant`, so the tangent at a point and the secant through two points are both one call.

## Derivative rules

Factor, sum/difference, product, chain, quotient — each checked against a `Symbolics`
derivative of the same expression.

## When differentiation fails

$\lvert x \rvert$ at $0$ (a kink: slope $+1$ from the right, $-1$ from the left) and
$\sqrt{x}$ at $0$ (a vertical tangent).

## Scratch

New helper functions worth keeping go in `src/` and get wired into `src/Calculus.jl` —
which currently has empty slots for *pure computational functions* and *integrated plotting
functions*, plus a commented-out `include`. With `Revise` loaded above, adding one there is

Figures follow the repo convention: build the plot into an explicit object and **return
it** (`plot!(plt, …)`, then `plt` as the last expression) rather than relying on

the docs alike.
